# Understanding ROS Nodes

 This tutorial introduces the use of [roscore](https://wiki.ros.org/roscore), [rosnode](https://wiki.ros.org/rosnode), and [rosrun](https://wiki.ros.org/rosrun) command-line tools.

## Following the tutorial

For this tutorial, we will need a graphical environment.
To launch it, you can run the following command in the terminal **of your laptop** (not the terminal in VSCode!):

    dts code vnc -R [!ROBOT_NAME]

This will spin up a container with ROS installed. 
A link will appear in the terminal that, when clicked, will open a lightweight VNC desktop environment:

## Nodes

A node really isn't much more than an executable file within a ROS package. ROS nodes use a ROS client library to communicate with other nodes. Nodes can publish or subscribe to a Topic. Nodes can also provide or use a Service. 

ROS client libraries allow nodes written in different programming languages to communicate:

- `rospy` : Python client library
- `roscpp` : C++ client library 

## roscore

roscore is the first thing you should run when using ROS. In our case, it is already running on the robot, so we don't need to start it.

## Using rosnode

Open up a **new terminal**, and let's use `rosnode` to see what nodes are currently running.

`rosnode` displays information about the ROS nodes that are currently running. The `rosnode list` command lists these active nodes:

    rosnode list

You will see a list of nodes that are already running on your robot, including `/rosout`, which is always running as it collects and logs nodes' debugging output, as well as 
nodes for the sensor and driver interfaces, which we use to read in sensor data and send actuation commands to make the robot take action. We don't need to worry about the details of most of them for now.

The `rosnode info` command returns information about a specific node.

    rosnode info /ROBOTNAME/camera_node

Where ROBOTNAME should be replaced by your robot name (the name you put after the `-R` flag when you ran the `dts code vnc` command above.
This gave us some more information about the camera node.

```bash
--------------------------------------------------------------------------------
Node [/ROBOTNAME/camera_node]
Publications: 
 * /rosout [rosgraph_msgs/Log]
 * /ROBOTNAME/camera_node/camera_info [sensor_msgs/CameraInfo]
 * /ROBOTNAME/camera_node/image/compressed [sensor_msgs/CompressedImage]

Subscriptions: None

Services: 
 * /vlx/camera_node/get_loggers
 * /vlx/camera_node/get_parameters_list
 * /vlx/camera_node/request_parameters_update
 * /vlx/camera_node/set_logger_level
 * /vlx/camera_node/switch


contacting node http://ROBOTNAME.local:33271/ ...
Pid: 5538
Connections: ...
```

**Note**: don't worry if the output that you get is not exactly the same as the above, but it should be generally similar. 

This tells us some things about this node, such that it _publishes_ data on three topics: 

 - `/rosout` which is of type [rosgraph_msgs/Log](https://docs.ros.org/en/noetic/api/rosgraph_msgs/html/msg/Log.html)
 - `/ROBOTNAME/camera_node/camera_info` which is of type [sensor_msgs/CameraInfo](https://docs.ros.org/en/noetic/api/sensor_msgs/html/msg/CameraInfo.html)
 - `/ROBOTNAME/camera_node/image/compressed` which is of type [/ROBOTNAME/camera_node/image/compressed](https://docs.ros.org/en/melodic/api/sensor_msgs/html/msg/CompressedImage.html)

 This node doesn't subscribe to anything (which is pretty typical for a sensor driver). Additionally, there are a few [services](https://wiki.ros.org/Services) which are advertised by this node. The details of these are not too important for now, but you can think of them as being specific functions that this node makes available, which can be called from anywhere.


## Using rosrun

`rosrun` allows you to use the package name to directly run a _node_ within a _package_ (without having to know the package path).

Usage:

    rosrun [package_name] [node_name]

So now we can run the `turtlesim_node` in the `turtlesim` package.

Then, in a new terminal:

    rosrun turtlesim turtlesim_node

You will see the turtlesim window:

![turtlesim](../assets/turtlesim.png) 

**NOTE**: The turtle may look different in your turtlesim window. Don't worry about it - there are many types of turtle and yours is a surprise!

In a new terminal run:

    rosnode list

You will see something that `turtlesim` is now included in the list.
One nice feature of ROS is that you can reassign Names from the command-line.

Close the turtlesim window to stop the node (or go back to the `rosrun turtlesim turtlesim_node` terminal and use ctrl-C). Now let's re-run it, but this time use a Remapping Argument to change the node's name:

    rosrun turtlesim turtlesim_node __name:=my_turtle

Now, if we go back and use `rosnode list`:

    rosnode list

We see our new `/my_turtle` node. Let's use another `rosnode` command, `ping`, to check that it's up:

    rosnode ping my_turtle

```bash
    rosnode: node is [/my_turtle]
    pinging /my_turtle with a timeout of 3.0s
    xmlrpc reply from http://aqy:42235/     time=1.152992ms
    xmlrpc reply from http://aqy:42235/     time=1.120090ms
    xmlrpc reply from http://aqy:42235/     time=1.700878ms
    xmlrpc reply from http://aqy:42235/     time=1.127958ms
```

## Review

What was covered:

- `roscore` = ros+core : master (provides name service for ROS) + rosout (stdout/stderr) + parameter server (parameter server will be introduced later)
- `rosnode` = ros+node : ROS tool to get information about a node.
- `rosrun` = ros+run : runs a node from a given package.
